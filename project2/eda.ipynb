# OMDB Data Exploratory Analysis

In this notebook, we will use the data collected to look at the distribution of movies by decade, the breakdown of movies vs series, and how many results each search term returned.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/processed/movies.csv")
print(f"{len(df)} total records")
df.head()
```

## Movies per Decade
How are the collected titles distributed across decades?

```python
df["decade"] = (df["year"] // 10 * 10).astype("Int64")
decade_counts = df["decade"].value_counts().sort_index()

decade_counts.plot(kind="bar", title="Movies per Decade", xlabel="Decade", ylabel="Count")
plt.tight_layout()
plt.show()
```

## Count by Type
Breakdown of movies vs series in the dataset.

```python
type_counts = df["Type"].value_counts()

type_counts.plot(kind="bar", title="Count by Type", xlabel="Type", ylabel="Count")
plt.tight_layout()
plt.show()
```

## Results per Search Term
How many results did each search term return?

```python
term_counts = df["search_term"].value_counts()

term_counts.plot(kind="bar", title="Results per Search Term", xlabel="Search Term", ylabel="Count")
plt.tight_layout()
plt.show()
```
